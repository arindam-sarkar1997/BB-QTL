# Fluconazole dose–response and a selected IC level

This notebook analyzes endpoint OD readings from the pooled HIP assay. Because the file contains concentrations rather than time points, the plotted curve is a **dose–response curve**. Set `IC_TARGET` in the next cell to `"IC20"`, `"IC40"`, `"IC50"`, or another percentage between 0 and 100, then run all cells. ICx means the fluconazole concentration at which blank-corrected OD is **(100 − x)% of the untreated control**.

Each `Rep` column is treated as one paired replicate across concentrations, its blank, and its untreated control. The plot shows mean ± SD of the normalized replicate readings and a 95% bootstrap confidence band for the fitted mean curve. The band and selected IC interval resample whole replicate columns so the within-replicate pairing is preserved.

The inhibitory Hill fit below has a free upper plateau and a lower plateau fixed at zero. Complete inhibition is not observed at the highest tested concentration, so the lower plateau is an assumption. Higher IC levels may require extrapolation beyond the tested concentrations; the notebook flags this when it occurs.

```python
from pathlib import Path
import re

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import least_squares
from IPython.display import display

DATA_PATH = Path("/blue/juannanzhou/arindam.sarkar/BB-QTL_HT/7_HIPHOP_validation/Pooled/MIC/FLC_MIC_0916.csv")
IC_TARGET = "IC50"  # Change to "IC40", "IC50", etc., then run all cells.
N_BOOT = 2000
RANDOM_SEED = 20260917

match = re.fullmatch(r"IC(\d+(?:\.\d+)?)", IC_TARGET.strip().upper()) if isinstance(IC_TARGET, str) else None
if match is None:
    raise ValueError('IC_TARGET must look like "IC20", "IC40", or "IC50".')
INHIBITION_PERCENT = float(match.group(1))
if not 0 < INHIBITION_PERCENT < 100:
    raise ValueError("The IC percentage must be greater than 0 and less than 100.")
IC_LABEL = f"IC{INHIBITION_PERCENT:g}"
TARGET_GROWTH = 1.0 - INHIBITION_PERCENT / 100.0

if not DATA_PATH.exists():
    raise FileNotFoundError(f"Set DATA_PATH to the location of FLC_MIC_0916.csv: {DATA_PATH}")
```

```python
raw = pd.read_csv(DATA_PATH, encoding="utf-8-sig")
label_col = raw.columns[0]
rep_cols = list(raw.columns[1:])
if len(rep_cols) < 2:
    raise ValueError("At least two replicate columns are needed for uncertainty estimates.")

labels = raw[label_col].astype(str).str.strip().str.lower()
blank_rows = labels.eq("blank")
control_rows = labels.eq("0")
if blank_rows.sum() != 1 or control_rows.sum() != 1:
    raise ValueError("Expected exactly one 'Blank' row and one '0' untreated-control row.")

od = raw[rep_cols].apply(pd.to_numeric, errors="coerce")
concentration = pd.to_numeric(raw[label_col], errors="coerce")
positive_rows = concentration.notna() & concentration.gt(0)
if positive_rows.sum() < 4:
    raise ValueError("At least four positive concentrations are needed for this fit.")
if od.loc[positive_rows | blank_rows | control_rows].isna().any().any():
    raise ValueError("OD values are missing or nonnumeric in rows used for the analysis.")

blank = od.loc[blank_rows].iloc[0].to_numpy(dtype=float)
control = od.loc[control_rows].iloc[0].to_numpy(dtype=float)
control_corrected = control - blank
if np.any(control_corrected <= 0):
    raise ValueError("Every untreated control must exceed its matching blank.")

order = np.argsort(concentration.loc[positive_rows].to_numpy(dtype=float))
doses = concentration.loc[positive_rows].to_numpy(dtype=float)[order]
if np.unique(doses).size != doses.size:
    raise ValueError("Positive concentration rows must be unique; combine duplicate rows first.")
od_positive = od.loc[positive_rows].to_numpy(dtype=float)[order]
growth = (od_positive - blank[None, :]) / control_corrected[None, :]

summary = pd.DataFrame({
    "FLC (µg/mL)": doses,
    "mean relative growth": growth.mean(axis=1),
    "SD across replicates": growth.std(axis=1, ddof=1),
    "n": growth.shape[1],
})
print(f"Loaded {growth.shape[1]} paired replicates at {growth.shape[0]} positive concentrations.")
display(summary)
```

```python
def hill_growth(x, top, scale, slope):
    # Zero lower asymptote; concentrations and scale have the same units.
    return top / (1.0 + (np.asarray(x) / scale) ** slope)


def fit_hill(observations, initial=None):
    if initial is None:
        initial = [1.0, float(np.median(doses)), 1.0]
    bounds = ([0.5, doses.min() / 1000, 0.1],
              [1.2, doses.max() * 1000, 5.0])
    result = least_squares(
        lambda p: (hill_growth(doses[:, None], *p) - observations).ravel(),
        x0=initial, bounds=bounds, max_nfev=5000,
    )
    if not result.success:
        raise RuntimeError(f"Hill fit failed: {result.message}")
    return result.x


def concentration_at_growth(params, target=TARGET_GROWTH):
    top, scale, slope = params
    if not 0 < target < top:
        raise ValueError(f"{IC_LABEL} is not reached: target growth exceeds the fitted upper plateau.")
    return scale * (top / target - 1.0) ** (1.0 / slope)


fit_params = fit_hill(growth)
ic_value = concentration_at_growth(fit_params)
print(f"Hill parameters: upper plateau = {fit_params[0]:.3f}, "
      f"scale = {fit_params[1]:.2f} µg/mL, slope = {fit_params[2]:.3f}")
print(f"Fitted {IC_LABEL} = {ic_value:.2f} µg/mL "
      f"({100 * TARGET_GROWTH:g}% of untreated growth)")
if not doses.min() <= ic_value <= doses.max():
    print(f"Caution: {IC_LABEL} lies outside the tested concentration range; "
          "this estimate is an extrapolation.")
```

```python
# Resample paired replicate columns, then refit each bootstrap sample.
rng = np.random.default_rng(RANDOM_SEED)
bootstrap_params = np.empty((N_BOOT, 3))
bootstrap_ic = np.full(N_BOOT, np.nan)

for b in range(N_BOOT):
    sampled_cols = rng.integers(0, growth.shape[1], size=growth.shape[1])
    params_b = fit_hill(growth[:, sampled_cols], initial=fit_params)
    bootstrap_params[b] = params_b
    if params_b[0] > TARGET_GROWTH:
        bootstrap_ic[b] = concentration_at_growth(params_b)

valid_ic = bootstrap_ic[np.isfinite(bootstrap_ic)]
if valid_ic.size == 0:
    raise RuntimeError(f"No bootstrap fits reach {IC_LABEL}; choose a higher inhibition level.")
if valid_ic.size < 0.95 * N_BOOT:
    print(f"Caution: {N_BOOT - valid_ic.size} bootstrap fits do not reach {IC_LABEL}; "
          "the percentile interval uses only the fits that do.")
ic_lower, ic_upper = np.quantile(valid_ic, [0.025, 0.975])
x_grid = np.geomspace(min(doses.min(), ic_value / 1.25),
                     max(doses.max(), ic_value * 1.25), 400)
bootstrap_curves = np.array([hill_growth(x_grid, *p) for p in bootstrap_params])
curve_lower, curve_upper = np.quantile(bootstrap_curves, [0.025, 0.975], axis=0)
print(f"{IC_LABEL} 95% bootstrap confidence interval: {ic_lower:.2f}–{ic_upper:.2f} µg/mL")
```

```python
fig, ax = plt.subplots(figsize=(9, 5.5), constrained_layout=True)
if ic_value > doses.max():
    ax.axvspan(doses.max(), x_grid.max(), color="#E5E7EB", alpha=0.6,
               label="Beyond tested concentrations")
elif ic_value < doses.min():
    ax.axvspan(x_grid.min(), doses.min(), color="#E5E7EB", alpha=0.6,
               label="Beyond tested concentrations")
ax.fill_between(x_grid, curve_lower, curve_upper, color="#3B82A0", alpha=0.22,
                label="Fitted curve: 95% bootstrap CI")
ax.plot(x_grid, hill_growth(x_grid, *fit_params), color="#17617C", lw=2.5,
        label="Fitted Hill curve")
ax.errorbar(doses, summary["mean relative growth"],
            yerr=summary["SD across replicates"], fmt="o", ms=5.5,
            capsize=3, color="#1F2937", ecolor="#64748B",
            label="Replicate mean ± SD")
ax.axhline(TARGET_GROWTH, color="#C35A32", ls="--", lw=1.3,
           label=f"{100 * TARGET_GROWTH:g}% of untreated control")
ax.axvline(ic_value, color="#C35A32", ls=":", lw=1.5)
ax.scatter([ic_value], [TARGET_GROWTH], s=75, facecolor="white",
           edgecolor="#C35A32", linewidth=2, zorder=5)
ax.annotate(f"{IC_LABEL} = {ic_value:.1f} µg/mL\n95% CI: {ic_lower:.1f}–{ic_upper:.1f}",
            xy=(ic_value, TARGET_GROWTH), xycoords="data",
            xytext=(0.53, 0.27 if TARGET_GROWTH >= 0.65 else 0.72),
            textcoords="axes fraction",
            arrowprops=dict(arrowstyle="->", color="#C35A32", lw=1.1),
            fontsize=10, color="#913F20",
            bbox=dict(boxstyle="round,pad=0.35", fc="white", ec="#E6B39F"))
ax.set_xscale("log", base=2)
ax.set_xlim(x_grid.min() / 1.1, x_grid.max() * 1.1)
shown_ticks = doses[np.unique(np.linspace(0, doses.size - 1, 6, dtype=int))]
ax.set_xticks(shown_ticks)
ax.set_xticklabels([f"{value:.3g}" for value in shown_ticks])
ax.set_ylim(min(0.35, TARGET_GROWTH - 0.1), 1.10)
ax.set_xlabel("Fluconazole (µg/mL)")
ax.set_ylabel("Relative growth (blank-corrected OD / untreated control)")
ax.set_title("Pooled HIP fluconazole dose–response")
ax.grid(True, which="major", alpha=0.18)
ax.legend(loc="lower left", frameon=True, fontsize=9)
plt.show()
```

**Reading the plot.** Error bars are the standard deviation of the eight normalized replicates at each concentration. The shaded area is a pointwise 95% confidence interval for the fitted mean curve from replicate bootstrap resampling; it is not a prediction interval for a new replicate. The selected IC interval uses the same bootstrap fits. This measures variability among the available replicates and does not account for differences between assay days or batches. Estimates beyond the tested concentrations depend more strongly on the assumed Hill curve.
