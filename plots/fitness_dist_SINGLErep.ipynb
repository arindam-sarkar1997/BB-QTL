```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# File paths
INPUT_TSV = "/blue/juannanzhou/arindam.sarkar/BB-QTL_HT/3_phenotype_inference/results/Independent_inference/single_rep/3tp/fitness_rep3_with_ids.tsv"
OUTPUT_FIG = "3tp_rep3.png"
```

```python
# Load TSV file
df = pd.read_csv(INPUT_TSV, sep=r'\s+', engine='python')

# Display summary statistics
print("Summary of fitness (s):")
print(df['s'].describe(), "\n")

# Show first few rows
display(df.head())
```

```python
sns.set(style="whitegrid", font_scale=1.2)
plt.figure(figsize=(8,5))
sns.histplot(df['s'], bins=50, kde=True, color='skyblue', edgecolor='black')

plt.title("Distribution of Fitness (s) — Replicate 3")
plt.xlabel("Fitness (s)")
plt.ylabel("Count")

plt.tight_layout()
plt.savefig(OUTPUT_FIG, dpi=300)
plt.show()
```
