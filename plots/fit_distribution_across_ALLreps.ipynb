```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/blue/juannanzhou/arindam.sarkar/BB-QTL_HT/3_phenotype_inference/results/Joint_inference/Unfiltered/CON/fitness_union_with_ids.tsv", sep="\t")
# Ensure numeric
for c in ["s","stderr(s)"]:
    df[c] = pd.to_numeric(df[c], errors="coerce")

print(df["s"].min(), df["s"].max())

plt.figure(figsize=(7,4))
plt.hist(df["s"].dropna(), bins=120, alpha=0.8)
plt.axvline(0, linestyle="--", linewidth=1)
plt.xlabel("Fitness (s)")
plt.ylabel("Number of lineages")
plt.title("Fluconazole - CON: Joint fitness across 3 UNFILTERED replicates and 3 timepoints (P0, P1 & P2)")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df["s"], df["stderr(s)"], s=6, alpha=0.5)
plt.xlabel("Fitness (s)")
plt.ylabel("Standard error of s")
plt.title("Fluconazole - CON: Uncertainty vs. estimated fitness")
plt.tight_layout()
plt.show()
```

```python
df_sorted = df.sort_values("s")
plt.figure(figsize=(7,4))
plt.plot(range(len(df_sorted)), df_sorted["s"], lw=0.8)
plt.xlabel("Lineages (sorted)")
plt.ylabel("Fitness (s)")
plt.title("Ranked fitness profile across ~100k lineages")
plt.tight_layout()
plt.show()
```
